```python
import os
import json

path = "/home/bjornwesterlun/IR_project"

with open(os.path.join(path, "../factscore_bio.jsonl"), "r", encoding="utf-8") as f:
    queries = [json.loads(line) for line in f]

with open(os.path.join(path, "data/query_retrieval.json"), "r", encoding="utf-8") as f:
    retrieval = json.loads(f.read())

print(len(queries))
print(len(retrieval))
queries = [query["prompt"] for query in queries]
# queries = queries[:5]  # Uncomment for quick testing
retrievals = [retrieval[x] for x in queries]
retrievals = [[doc for doc in retrieval[x]] for x in queries]

full_prompts = []
n = 1
for query, retrievals in zip(queries[:n], retrievals[:n]):
    full_prompt = f"Retrieved context:\n"
    for i, retrieval in enumerate(retrievals):
        title = retrieval["title"]
        content = retrieval["contents"]
        full_prompt += f"Document {i} with title {title}:\n{content}\n"
    full_prompt += f"Question to answer: {query}"
    full_prompts.append(full_prompt)
```

```python
from transformers import AutoTokenizer

local_model_path = "/vol/csedu-nobackup/course/I00041_informationretrieval/users/bjorn/Qwen2.5-7B"
tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-7B", cache_dir=local_model_path,)

to_tokenizer = [[{"role": "system", "content": "You are a helpful assistant."}, {"role": "user", "content": full_prompt}] for full_prompt in full_prompts]

tokenized = tokenizer.apply_chat_template(to_tokenizer, return_tensors="pt", padding=True)
```

```python
conversation = [{"role": "system", "content": "You are a helpful assistant."}, {"role": "user", "content": full_prompt}]
result = tokenizer.apply_chat_template(conversation, tokenize = False)
print(result)
```
