```python
import graphviz
relevant = ["Kaoru Kuroki has appeared in Japanese television dramas.", "Kaoru Kuroki has appeared in television dramas.", "Kaoru Kuroki has worked in film and television.", "Kaoru Kuroki is a multi-media personality.",]# "Kaoru Kuroki became a media counselor.",] #"Kaoru Kuroki is a media personality."]
def visualize_hierarchy(matrix, clusters):
    dot = graphviz.Digraph(comment='My Hierarchical DAG', graph_attr={'rankdir': 'TB'})
    representatives = [c["representative"] for c in clusters]
    for i in range(len(representatives)):
        if representatives[i]["atomic_fact"] not in relevant:
            continue
        dot.node(f'{representatives[i]["atomic_fact"]}', f'{representatives[i]["atomic_fact"]}')
        for j in range(len(representatives)):
            if i == j:
                continue
            elif matrix[i][j][2] > 0.65:
                rep = representatives[i]
                other_rep = representatives[j]

                dot.edge(f'{rep["atomic_fact"]}', f'{other_rep["atomic_fact"]}', label=f'{matrix[i][j][2]:.2f} ({matrix[j][i][2]:.2f})')
    return dot
n = 585
dot = visualize_hierarchy(matrices[n], cluster_data[n]["clusters"])

dot.attr(dpi="700")
# create png
dot.render('hierarchy', format='png')
```
